 Q5. Identify the most frequently appearing `item_name`

```python
# value_counts() returns item_name frequencies in descending order.
# idxmax() retrieves the index label (item name) of the maximum count.
most_frequent_item = df['item_name'].value_counts().idxmax()
most_frequent_count = df['item_name'].value_counts().max()

print(f"Most frequent item: '{most_frequent_item}' — appears {most_frequent_count} times")
```

Q8. Convert `item_price` from string to float

```python
#Created a helper function to convert the string prices to float
def parse_price(price_str):
    """
    Convert a price string like '$X.XX' to a float

    Parameters
    ----------
    price_str : str
        A price string that begins with a dollar sign, e.g '$2.39'

    Returns
    -------
    float
        The numeric price value, e.g. 2.39.
    """
    # Remove the leading '$', then turn to float
    return float(price_str.replace('$', '').strip())


# Apply the helper function to all values in the item_price column
# We then overwrite the column in place so the rest of the notebook
# can rely on item_price being a float type
df['item_price'] = df['item_price'].apply(parse_price)

# Confirm if the conversion was successful
print(f"item_price dtype after conversion: {df['item_price'].dtype}")
print("\nSample values:")
print(df['item_price'].head())
```

Q9. Calculate the average `item_price` across all orders

```python
# mean() calculates the average of all non-null numeric values in the df
# Now that item_price is a float, this works directly without needing
# to convert further
average_price = df['item_price'].mean()

print(f"Average item price: ${average_price:.2f}")
```

Q10. Determine the total number of unique orders

```python
# nunique() on the order_id column gives us the count of distinct order IDs,
# which are the total number of unique orders placed.
unique_orders = df['order_id'].nunique()

print(f"Total number of unique orders: {unique_orders}")
```

## Section 2: Extra Credit — Total Revenue Calculation (Pandas)

Q1. Calculate Total Revenue

Each row in the dataset represents an item in an order
Revenue formula used = `quantity × item_price` 
Total revenue = sum of revenue across all items

```python
# Frst, create a 'line_revenue' column for each row.
# This multiplies the number of units ordered by the unit price
df['line_revenue'] = df['quantity'] * df['item_price']

# Then, sum all line revenues to get the total revenue across all orders.
total_revenue = df['line_revenue'].sum()

print(f"Total Revenue from all orders: ${total_revenue:,.2f}")
```
